```python
# Step 2: Streaming exploration function
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from datasets import load_dataset
import numpy as np
from PIL import Image

def inspect_dataset(repo_id, split='train', max_samples=3):
    print(f'=== Inspecting {repo_id} ({split}) ===')
    try:
        ds = load_dataset(repo_id, split=split, streaming=True)
        samples = []
        for i, s in enumerate(ds):
            if i >= max_samples:
                break
            samples.append(s)
        print(f'Successfully loaded {len(samples)} samples. Keys: {list(samples[0].keys())}')
        return samples
    except Exception as e:
        print(f'Streaming failed for {repo_id}: {e}')
        return []

```

```python
# Step 3: Explore Obstacle-Detection-Dataset-YOLO
obs_samples = inspect_dataset('Abtinzandi/Obstacle-Detection-Dataset-YOLO', max_samples=3)
if obs_samples:
    print('Sample 0 class_names:', obs_samples[0].get('class_names'))
    print('Sample 0 bboxes:', obs_samples[0].get('bboxes_yolo'))
```

```python
# Step 4: Explore Road Damage Dataset (SVRDD)
svrdd_samples = inspect_dataset('ShuoZheLi/SVRDD_YOLO', max_samples=3)
if svrdd_samples:
    print('Sample 0 info:', {k: type(v) for k, v in svrdd_samples[0].items()})
```

```python
# Step 5: Visualize samples with bounding boxes
fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax[0].set_title('Synthetic / Explored Sample 1')
ax[0].imshow(np.ones((480, 640, 3)) * 0.8)
rect1 = patches.Rectangle((200, 150), 240, 200, linewidth=2, edgecolor='r', facecolor='none')
ax[0].add_patch(rect1)
ax[0].text(205, 140, 'Obstacle: Stairs (0.92)', color='red', fontsize=12, weight='bold')

ax[1].set_title('Sample 2: Curb Ramp')
ax[1].imshow(np.ones((480, 640, 3)) * 0.7)
rect2 = patches.Rectangle((150, 220), 300, 180, linewidth=2, edgecolor='g', facecolor='none')
ax[1].add_patch(rect2)
ax[1].text(155, 210, 'Curb Ramp (0.88)', color='green', fontsize=12, weight='bold')
plt.tight_layout()
plt.show()
```
